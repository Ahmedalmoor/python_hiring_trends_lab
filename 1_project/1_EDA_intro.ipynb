### 1. Import Libraries, Load and Clean Data

```python
import ast
import pandas as pd
import seaborn as sns
from datasets import load_dataset
import matplotlib.pyplot as plt


dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()


df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast. literal_eval(x) if pd.notna(x) else x)
```

### 2. Most Popular Jobs

```python
df_plot = df['job_title_short'].value_counts().to_frame()

sns.set_theme(style='ticks')
sns.barplot(
    data=df_plot,
    x='count',
    y='job_title_short',
    hue='count',
    palette='dark:b_r',
    legend=False
)
plt.title('Number of Jobs per Job Title')
plt.xlabel('Number of Jobs')
plt.ylabel('')
plt.show()
```

### 3. Countries to Explore

```python
df_plot = df['job_country'].value_counts().head(20).to_frame()

sns.set_theme(style='ticks')
sns.barplot(
    data=df_plot,
    x='count',
    y='job_country',
    hue='count',
    palette='dark:b_r',
    legend=False
)
plt.title('Number of Jobs per Country')
plt.xlabel('Number of Jobs')
plt.ylabel('')
plt.show()
```

### 4. Companies to Explore

```python
df_plot = df['company_name'].value_counts().head(20).to_frame()

sns.set_theme(style='ticks')
sns.barplot(
    data=df_plot,
    x='count',
    y='company_name',
    hue='count',
    palette='dark:b_r',
    legend=False
)
plt.title('Number of Jobs per Company')
plt.xlabel('Number of Jobs')
plt.ylabel('')
plt.show()
```

### 5. Job Opportunities (Percentages)

```python
dict_column = {
    'job_work_from_home': 'Work from Home Offered',
    'job_no_degree_mention': 'Degree Requirement',
    'job_health_insurance': 'Health Insurance Offered'
}

fig, ax = plt.subplots(1, 3, figsize=(12, 5))
for i, (column, title) in enumerate(dict_column.items()):
    df_plot = df[column].value_counts(normalize=True).to_frame() * 100
    sns.barplot(
        data=df_plot,
        x=df_plot.index,
        y='proportion',
        ax=ax[i],
        palette='dark:b_r',
        legend=False
    )
    ax[i].set_title(title)
    ax[i].set_xlabel('')
    ax[i].set_ylabel('')
    ax[i].yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: f'{int(y)}%'))
plt.show()
```

### 6. Filter for Data Analyst Roles (Global)

```python
df_analyst = df[df['job_title_short'] == 'Data Analyst']
```

### 7. Locations to Explore for Data Analyst (Global)

```python
sns.set_theme(style='ticks')
sns.barplot(
    data=df_analyst['job_location'].value_counts().head(10).to_frame(),
    x='count',
    y='job_location',
    hue='count',
    palette='dark:b_r',
    legend=False
)
plt.title('Counts of Job Locations for Data Analyst')
plt.xlabel('Number of Jobs')
plt.ylabel('')
plt.show()
```

### 8. Job Opportunities for Data Analyst (Global - Percentages)

```python
fig, ax = plt.subplots(1, 3, figsize=(12, 5))
for i, (column, title) in enumerate(dict_column.items()):
    df_plot = df_analyst[column].value_counts(normalize=True).to_frame() * 100
    sns.barplot(
        data=df_plot,
        x=df_plot.index,
        y='proportion',
        ax=ax[i],
        palette='dark:b_r',
        legend=False
    )
    ax[i].set_title(title)
    ax[i].set_xlabel('')
    ax[i].set_ylabel('')
    ax[i].yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: f'{int(y)}%'))
plt.show()
```
